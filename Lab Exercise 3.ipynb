```python
import numpy as np
from numpy.linalg import inv
from numpy.linalg import det

# 1. Define matrices A, B, and C
A = np.array([[5, 2, 3],
              [6, 1, 3],
              [2, 4, 1]])
print("Matrix A:")
print(A)

B = np.array([[5, 2, 3],
              [6, 1, 3],
              [2, 4, 1]])
print("\nMatrix B:")
print(B)

C = np.array([[5, 2, 3],
              [6, 1, 3],
              [2, 4, 1]])
print("\nMatrix C:")
print(C)

# 2. Define matrix D
D = np.array([[4, -7, 4],
              [5, 6, -1],
              [-4, 2, 5]])
print("\nMatrix D:")
print(D)

# 3. Calculate AD and DA
AD = np.matmul(A, D)
DA = np.matmul(D, A)

print("\nAD:")
print(AD)
print("\nDA:")
print(DA)
print("\nAre AD and DA equal?", np.array_equal(AD, DA))

# 4. Calculate determinant and inverse of B
print("\nDeterminant of B:")
print(det(B))

print("\nInverse of B:")
B_inv = inv(B)
print(B_inv)

# 5. Calculate inverse of D and verify DD^(-1) and D^(-1)D
D_inv = inv(D)
print("\nInverse of D:")
print(D_inv)

DD_inv = np.matmul(D, D_inv)
D_invD = np.matmul(D_inv, D)

print("\nDD^(-1):")
print(DD_inv)
print("\nD^(-1)D:")
print(D_invD)

# Observation: Both DD^(-1) and D^(-1)D are identity matrices
# (within numerical precision)
```
