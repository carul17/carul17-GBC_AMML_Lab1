```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

# # of hours a AI student spends on their courses in a month
np.random.seed(10)
no_hours = np.random.normal(40.0, 10.0, 10000)

plt.hist(no_hours, 50)
plt.show()
```

```python
# Calculate and report the mean, std, and var of the data shown above.
# What is the theoretical relationship between std and var? Do the numbers reperted in the previous step agree with your expectation?
# Begin changing the different parameters on the normal function, discuss your observations as you change each input parameter.

# Calculate statistics
mean = np.mean(no_hours)
std = np.std(no_hours)
var = np.var(no_hours)

print(f"Mean: {mean:.2f}")
print(f"Standard Deviation: {std:.2f}")
print(f"Variance: {var:.2f}")

# The theoretical relationship between std and var is:
# var = std^2
# to verify this: 
print(f"\nVerification - std^2: {std**2:.2f}")
# The numbers agree with the theoretical relationship, as variance is approximately equal to standard deviation squared
```

```python
# 1. Lower mean
data_low_mean = np.random.normal(20.0, 10.0, 10000)
plt.hist(data_low_mean, 50)
plt.title('Lower Mean Distribution (mean=20.0, std=10.0)')
plt.show()
```

```python
# 2. Higher mean
data_high_mean = np.random.normal(60.0, 10.0, 10000)
plt.hist(data_high_mean, 50)
plt.title('Higher Mean Distribution (mean=60.0, std=10.0)')
plt.show()
```

```python
# 3. Lower standard deviation
data_low_std = np.random.normal(40.0, 5.0, 10000)
plt.hist(data_low_std, 50)
plt.title('Lower Std Distribution (mean=40.0, std=5.0)')
plt.show()
```

```python
# 4. Higher standard deviation
data_high_std = np.random.normal(40.0, 20.0, 10000)
plt.hist(data_high_std, 50)
plt.title('Higher Std Distribution (mean=40.0, std=20.0)')
plt.show()
```

```python
# Observations:
print("\nObservations:")
print("1. Changing the mean shifts the center of the distribution:")
print("   - Lower mean (20.0) moves the peak to the left")
print("   - Higher mean (60.0) moves the peak to the right")
print("\n2. Changing the standard deviation affects the spread:")
print("   - Lower std (5.0) makes the distribution more peaked and narrow")
print("   - Higher std (20.0) makes the distribution more spread out and flatter")
```
